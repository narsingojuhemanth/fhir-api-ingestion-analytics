# Initialize the FHIR assessment Lakehouse

Attach lh_fhir_assignment as the default Lakehouse. Ensure the Lakehouse was created with schema support enabled. Run this notebook once before the pipeline; it creates the three Medallion schemas and does not delete or overwrite any table.

```python
# Create the Medallion schemas in the attached, schema-enabled Lakehouse.
# This is safe to execute repeatedly and does not delete data.
for schema_name in ["bronze", "silver", "gold"]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {schema_name}")

print("Lakehouse schemas ready: bronze, silver, gold")
display(spark.sql("SHOW SCHEMAS"))
```
